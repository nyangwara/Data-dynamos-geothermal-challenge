# Reservoir deliverability calculation


Uses Darcy radial flow + USP-01 properties to verify the predicted flow rate is achievable at acceptable wellhead pressures.

## Inputs
- BLT-01 ThermoGIS values (P50)
- USP-01 predicted properties (IDW from BLT and JUT)

## Outputs
- Flow rate vs. pressure drop curve
- Validation that 105 m³/h is achievable


```python
import numpy as np
import matplotlib.pyplot as plt

# Reservoir parameters
k = 75e-15       # m² (75 mD)
h = 123          # m thickness
mu = 0.0005      # Pa·s (water at 76°C)
phi = 0.114      # porosity

# Geometry
rw = 0.108       # wellbore radius (8.5-inch hole)
re = 1500        # drainage radius (1.5 km, matches doublet spacing)

# Darcy radial flow: q = 2*pi*k*h*dP / (mu * ln(re/rw))
def flow_rate_m3s(dP_bar):
    dP_Pa = dP_bar * 1e5
    q = 2 * np.pi * k * h * dP_Pa / (mu * np.log(re/rw))
    return q  # m³/s

dPs = np.linspace(0, 100, 50)
qs = [flow_rate_m3s(dp) * 3600 for dp in dPs]  # m³/h

target = 105
target_dp = target / 3600 / (2*np.pi*k*h / (mu*np.log(re/rw))) / 1e5
print(f"Drawdown required for 105 m³/h: {target_dp:.1f} bar")
print(f"Typical safe drawdown for NL Slochteren: 50-80 bar → 105 m³/h is comfortable")
```

## Conclusion

The Darcy radial-flow calculation above returns a required drawdown of ~24 bar for 105 m³/h (see the printed value), well within the safe operating envelope for the Slochteren reservoir (typical 50–80 bar). The thermal-breakthrough constraint (1.5 km spacing) is the binding design constraint, not pressure.
