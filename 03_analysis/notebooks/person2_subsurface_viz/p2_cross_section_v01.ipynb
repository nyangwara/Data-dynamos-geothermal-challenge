# Geological cross-section — SW-NE transect


Builds the NE-SW geological cross-section through the four wells plus the proposed USP-01 location. The Slochteren reservoir top dips ~200 m from JUT-01 (NE) to BLT-01 (SW).

## Output
- `05_visualizations/cross_sections/cross_section_sw_ne_v01.png`


```python
import sys
sys.path.insert(0, '../../../03_analysis/scripts')
import matplotlib.pyplot as plt
import numpy as np
from plotting_utils import apply_style, COLORS
apply_style()

# (easting, top_tvd, base_tvd, phie%) projected onto the SW->NE (easting) axis
wells = {'JUT-01': (134098, 1655, 1781, 10.9),
         'EVD-01': (136997, 1783, 1860, 8.0),
         'USP-01': (141278, 1830, 1953, 11.4),
         'BLT-01': (141577, 1862, 1985, 11.4)}
order = sorted(wells, key=lambda w: wells[w][0])
xs    = np.array([wells[w][0] for w in order])
tops  = np.array([wells[w][1] for w in order])
bases = np.array([wells[w][2] for w in order])

fig, ax = plt.subplots(figsize=(11, 7))
ax.fill_between(xs, tops, bases, color=COLORS['sand'], alpha=0.55, label='Slochteren reservoir')
ax.plot(xs, tops,  '-o', color=COLORS['producer'], lw=1.6, label='Reservoir top (TVD)')
ax.plot(xs, bases, '-o', color=COLORS['shale'],    lw=1.2, label='Reservoir base (TVD)')
for w in order:
    x, t, b, p = wells[w]
    col = COLORS['injector'] if w == 'USP-01' else 'black'
    ax.plot([x, x], [t - 15, b + 15], color=col, lw=0.9, ls=':')
    ax.annotate(f"{w}\n\u03d5e {p}%", (x, t), textcoords='offset points', xytext=(0, -30),
                ha='center', fontsize=9, fontweight='bold', color=col)
ax.invert_yaxis()
ax.set_xlabel('Easting (RD New, m)   \u2014   SW \u2192 NE')
ax.set_ylabel('TVD (m)')
ax.set_title('SW\u2013NE geological cross-section through the doublet fairway', loc='left', fontweight='bold')
ax.legend(loc='lower left')
ax.text(0.99, 0.04, 'PKP-01 (far W, 2207 m TVD) is off-section \u2014 interpreted separate fault block',
        transform=ax.transAxes, ha='right', fontsize=8, style='italic', color=COLORS['bad'])
plt.tight_layout()
plt.savefig('../../../05_visualizations/cross_sections/cross_section_sw_ne_v01.png', dpi=150, bbox_inches='tight')
plt.savefig('../../../05_visualizations/cross_sections/cross_section_sw_ne_v01.svg', bbox_inches='tight')
plt.show()
```

## Geological interpretation

The transect shows three observations:

1. **Slochteren reservoir dips from NE to SW** — top is at 1655 m TVD at JUT-01, deepens to 1862 m at BLT-01. This is consistent with the regional structural model (NL onshore Permian generally dips north).
2. **PKP-01 is off-trend** — at 2207 m TVD (22 km west of USP), it's deeper than expected. Either the basin deepens westward or PKP-01 sits in a separate fault block. The tight, shaly reservoir at this location supports the latter hypothesis.
3. **USP-01 (proposed) is interpolated** — at 1830 m TVD, between BLT-01 (1862) and JUT-01 (1655). The reservoir thickness should remain ~120 m consistent with the BLT-01 analog.

The doublet plan exploits the structurally simple area between BLT-01 and the USP, avoiding the deeper/faulted region to the west (EVD-01, PKP-01).
